### 3.3 if / elif / else

*Slide 15 - 20 minutes*

Many AI systems make decisions with exactly the structure you are about to
write. The same logic that flags a "high-risk loan applicant" can reproduce
human bias.

Python checks the conditions from top to bottom and runs the first one that is
True. Indentation is not styling in Python - it is the syntax that decides
which lines belong to which branch.

- Create the variable `income` with `income = int(input('Annual income in euros: '))`
- Write an `if` / `elif` / `else` block that assigns a risk level:
  - `income` below 20000 gives `'high risk'`
  - `income` below 40000 gives `'medium risk'`
  - anything else gives `'low risk'`
- Print: `Based on income alone, this person is classified as: [risk level]`
- Then also ask for `age`, and combine the two with `and` or `or` so that the
  classification uses both.

````python
income = int(input('Annual income in euros: '))

if income < 20000:
    risk = "high risk"
elif income < 40000:
    risk = "medium risk"
else:
    risk = "low risk"

print("Based on income alone, this person is classified as: " + risk)

age = int(input('Age in numbers: '))

if income < 20000 or age < 18:
    risk = "high risk"
elif income < 40000 or (18 <= age < 30):
    risk = "medium risk"
else:
    risk = "low risk"

print("Based on income and age, this person is classified as: " + risk)
````

## 7. Five things to remember

   Use brackets when in doubt.

   before you calculate with it.
4. Indentation is syntax, not style. Four spaces decide which lines belong to
   which branch.

   tomorrow. Design the conditions carefully.
